```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce

# Task 1: Create a DataFrame using Pandas
# Create a DataFrame with the following data:
# Name: Alice, Bob, Charlie, David
# Age: 25, 30, 35, 28
# City: New York, San Francisco, Los Angeles, Chicago

# Your code here:
# Task 1: Create a DataFrame with Name, Age, and City
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
}

# Convert dictionary to a Pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame and data types to confirm correctness
print("--- Task 1 Output ---")
print(df)
print("\nData Types:")
print(df.dtypes)


# Task 2: Row and Column Manipulation
# Drop the 'City' column from the DataFrame created in Task 1

# Your code here:
# Task 2: Drop the 'City' column from the DataFrame created in Task 1
# 'axis=1' specifies columns, and 'inplace=False' returns a modified copy safely
df_dropped = df.drop('City', axis=1)

# Verify that the resulting DataFrame only has 'Name' and 'Age' columns
print("--- Task 2 Output ---")
print(df_dropped)


# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them

# Your code here:
# Task 3: Create a new DataFrame containing null/missing values
null_data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Tablet'],
    'Price': [1200.0, 25.0, None, 300.0, None],
    'Stock': [10, None, 50, None, 15]
}
df_null = pd.DataFrame(null_data)
print("--- Original DataFrame with Nulls ---")
print(df_null)

# Fill null values: use 0 for numeric columns ('Stock'), and a placeholder for 'Price'
df_filled = df_null.fillna({'Price': df_null['Price'].mean(), 'Stock': 0})

print("\n--- Filled DataFrame Output ---")
print("--- Task 3 Output ---")
print(df_filled)


# Task 4: GroupBy and Describe
# Using the following DataFrame, group by 'Category' and describe the 'Value' column

df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

# Your code here:
# Task 4: Use GroupBy to group data by 'Category' and apply describe() to 'Value'
grouped = df_group.groupby('Category')['Value']

# Calculate descriptive statistics for each category
describe_output = grouped.describe()

print("--- Task 4 Output ---")
print(describe_output)

# Task 5: Concatenation and Merging
# Concatenate two DataFrames vertically and merge them horizontally

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Your code here:
# Task 5: Concatenate df1 and df2 vertically, then merge the result with df3 horizontally
# Concatenating df1 and df2 vertically (axis=0)
vertical_concat = pd.concat([df1, df2], axis=0, ignore_index=True)

# Merging / Joining vertically combined data with df3 horizontally (axis=1)
final_merged = pd.concat([vertical_concat, df3], axis=1)

print("--- Task 5 Output ---")
print(final_merged)
print(f"\nFinal Shape: {final_merged.shape}")

# Task 6: Tuples and Sets
# Create a tuple of fruits and a set of numbers.
# Then, try to add an element to each and observe the difference.

# Your code here:
# Task 6: Tuples vs Sets demonstration
my_tuple = ('Apple', 'Banana', 'Orange')
my_set = {1, 2, 3, 4, 5}

# Attempting to add an element to both structures
try:
    # Tuples are immutable and do not support item addition/mutation
    my_tuple.append('Grapes')
except AttributeError as e:
    print(f"Tuple Error Handled: {e}")

# Sets are mutable and unordered; add() is valid
my_set.add(6)

print("--- Task 6 Output ---")
print(f"Tuple (unchanged): {my_tuple}")
print(f"Set (modified with new element): {my_set}")

# Task 7: Dictionaries
# Create a dictionary of student names and their scores.
# Then, update a student's score and add a new student.

# Your code here:
# Task 7: Create a dictionary of student scores, update one, and add a new student
student_scores = {
    'Alice': 85,
    'Bob': 90,
    'Charlie': 78
}

# Update the score of an existing student (Bob)
student_scores['Bob'] = 95

# Add a new student to the dictionary
student_scores['David'] = 88

print("--- Task 7 Output ---")
print(student_scores)


# Task 8: Functions and Lambda
# Create a function to calculate the square of a number.
# Then, use a lambda function to do the same.

# Your code here:
# Task 8: Regular function vs Lambda function to calculate the square of a number
def square_regular(x):
    """Calculates the square of a number using a standard function."""
    return x ** 2

# Equivalent anonymous function using lambda
square_lambda = lambda x: x ** 2

# Demonstrate with two different inputs (e.g., 4 and 7)
inputs = [4, 7]
print("--- Task 8 Output ---")
for val in inputs:
    print(f"Input: {val} | Regular Function: {square_regular(val)} | Lambda Function: {square_lambda(val)}")

# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
# Task 9: Iterator class and Generator function for the first 5 even numbers (2, 4, 6, 8, 10)

# 1. Custom Iterator Class
class EvenNumbersIterator:
    def __init__(self, limit=5):
        self.limit = limit
        self.current = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.limit:
            raise StopIteration
        even_val = self.current * 2
        self.current += 1
        self.count += 1
        return even_val

# 2. Generator Function using 'yield'
def even_numbers_generator(limit=5):
    for i in range(1, limit + 1):
        yield i * 2

print("--- Task 9 Output ---")
print("Iterator Output:", list(EvenNumbersIterator(5)))
print("Generator Output:", list(even_numbers_generator(5)))

# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
squared_numbers = list(map(lambda x: x ** 2, numbers))
product_numbers = reduce(lambda x, y: x * y, numbers)
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

# Print the results of each operation

print("--- Task 10 Output ---")
print("Squared numbers:", squared_numbers)
print("Product of numbers:", product_numbers)
print("Even numbers:", even_numbers)


# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)

# Demonstration of the Rectangle class
my_rectangle = Rectangle(10, 5)
print("--- Task 11 Output ---")
print("Rectangle Area:", my_rectangle.calculate_area())
print("Rectangle Perimeter:", my_rectangle.calculate_perimeter())

# Task 12: Pandas Data Analysis
# Using the following DataFrame, perform these tasks:
# 1. Find the average salary by department
# 2. Get the names of employees with salary > 60000
# 3. Add a new column 'Bonus' which is 10% of salary

df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# Your code here:
# 1. Find the average salary by department
avg_salary_by_dept = df_employees.groupby('Department')['Salary'].mean()

print("--- Task 12 Output ---")
print("1. Average Salary by Department:\n", avg_salary_by_dept)

# 2. Get the names of employees with salary > 60000
high_salary_names = df_employees[df_employees['Salary'] > 60000]['Name']
print("\n2. Employees with Salary > 60000:\n", high_salary_names.tolist())

# 3. Add a new column 'Bonus' which is 10% of salary
df_employees['Bonus'] = df_employees['Salary'] * 0.10
print("\n3. Updated DataFrame with Bonus:\n", df_employees)

```
